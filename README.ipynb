# Airline Passenger Satisfaction Analysis

## Project Overview

This project analyzes airline passenger data to understand the factors associated with passenger satisfaction.

The analysis follows the **CRISP-DM (Cross-Industry Standard Process for Data Mining)** framework, covering:

1. Business Understanding
2. Data Understanding
3. Data Preparation
4. Modeling
5. Evaluation
6. Deployment and Recommendations

The goal is to identify areas where an airline can improve the passenger experience and use machine learning to predict passenger satisfaction.

### **CRISP DM Methodology**

The project will follow the CRISP-DM framework.

### 1. Business Understanding

Define the business problem, objectives, and questions that the analysis will address.

### 2. Data Understanding

Explore the dataset, examine its structure, identify missing values and duplicates, and investigate important patterns and relationships.

### 3. Data Preparation

Clean the data, handle missing values, remove unnecessary variables, encode categorical variables, engineer useful features, and prepare the data for machine learning.

### 4. Modeling

Develop classification models to predict passenger satisfaction.

Potential models include:

- Logistic Regression
- Decision Tree
- Random Forest

### 5. Evaluation

Evaluate model performance using appropriate classification metrics, including:

- Accuracy
- Precision
- Recall
- F1-score
- Confusion Matrix

### 6. Deployment and Recommendations

Translate the analysis and model results into practical recommendations that can help the airline improve passenger satisfaction.

# 3. Data Preparation

Data preparation is the process of getting the dataset ready for analysis and machine learning.

In this stage, we will:

1. Remove unnecessary columns.
2. Check and handle missing values.
3. Check categorical variables.
4. Prepare the target variable.
5. Separate the features from the target.
6. Split the data into training and testing sets.
7. Prepare the data for machine-learning models.

The goal is to make sure the data is clean, consistent, and in a suitable format for modeling.

### 3.1 Remove Unnecessary Columns

The `Unnamed: 0` and `id` columns were removed because they are identifiers and do not provide meaningful information about passenger satisfaction.

### 3.2 Check Missing Values

The dataset contained 83 missing values in `Arrival Delay in Minutes`, representing approximately 0.32% of the 25,976 passenger records.

### 3.3 Handling Missing Values

The missing `Arrival Delay in Minutes` values were replaced using the median.

The median was selected because the delay data is right-skewed and contains some extreme delay values. The median is less affected by extreme values than the mean.

### 3.4 Check Categorical Variables

The categorical variables were checked for their unique values.

The categorical features were:

- `Gender`
- `Customer Type`
- `Type of Travel`
- `Class`

The categories were consistent and suitable for encoding.

### 3.5 Prepare Target Variable

The target variable, `satisfaction`, was converted from text into numerical values:

- `1` = Satisfied
- `0` = Neutral or dissatisfied

This allows the problem to be treated as a binary classification task.

### 3.6 Separate Features and Target

The dataset was divided into:

- `X` — predictor variables
- `y` — target variable

The features contained 25,976 passenger records and 22 predictor variables.

### 3.7 Split Data into Training and Testing Sets

The data was divided into:

- 80% training data
- 20% testing data

This resulted in:

- Training data: 20,780 records
- Testing data: 5,196 records

A `random_state` of 42 was used to make the split reproducible.

Stratification was also used to maintain a similar satisfaction distribution in both datasets.

### 3.8 Check Satisfaction Distribution After Split

The satisfaction distribution remained almost identical between the training and testing datasets.

- Training: 56.10% neutral/dissatisfied and 43.90% satisfied
- Testing: 56.10% neutral/dissatisfied and 43.90% satisfied

This confirms that stratification worked successfully.

### 3.9 Encode Categorical Variables

One-Hot Encoding was used to convert categorical variables into numerical features.

The 4 categorical features were transformed while the 18 numerical features were retained.

After encoding, the dataset contained 23 processed features.

### 3.10 Apply the Preprocessing

A `ColumnTransformer` was used to apply different preprocessing steps to numerical and categorical variables.

- Numerical variables were standardized using `StandardScaler`.
- Categorical variables were converted using `OneHotEncoder`.

The preprocessing was incorporated into a machine-learning pipeline.

The processed datasets contained:

- Training: 20,780 records × 23 features
- Testing: 5,196 records × 23 features

# 4. Predictive Modeling

The predictive modeling stage investigates whether passenger satisfaction can be predicted using passenger, travel, service, and delay information.

The target is a binary classification problem:

- `1` = Satisfied
- `0` = Neutral or dissatisfied

## 4.1 Logistic Regression

Logistic Regression was selected as the first classification model because it is suitable for predicting an outcome with two categories.

The numerical features were scaled before training because the variables have very different ranges.

The model was trained using the training dataset and evaluated using the unseen testing dataset.

## 4.2 Model Predictions

The trained Logistic Regression model was used to predict satisfaction outcomes for the testing dataset.

The predictions were stored in `y_pred`.

## 4.3 Logistic Regression Performance

The Logistic Regression model achieved:

| Metric | Score |
|---|---:|
| Accuracy | 87.24% |
| Precision | 86.18% |
| Recall | 84.48% |
| F1-score | 85.32% |

### Key Finding

The model achieved an overall accuracy of 87.24%, indicating that it correctly classified approximately 87 out of every 100 passengers in the testing dataset.

The precision of 86.18% indicates that most passengers predicted as satisfied were actually satisfied.

The recall of 84.48% indicates that the model identified most of the passengers who were actually satisfied.

The F1-score of 85.32% indicates a good balance between precision and recall.

### Business Interpretation

The results suggest that passenger characteristics, travel information, service ratings, and flight delays contain useful information for predicting passenger satisfaction.

However, the model is not perfect and makes some incorrect predictions. Further evaluation using a confusion matrix and comparison with other classification models will help determine whether the predictive performance can be improved.



> **Important:** Predictive performance shows that variables are useful for prediction. It does not prove that these variables directly cause passenger satisfaction or dissatisfaction.

## 4.10 Confusion Matrix for the Final Model

The Decision Tree was selected as the final model because it performed better than Logistic Regression across all evaluation metrics.

The confusion matrix was used to understand how well the Decision Tree classified passengers into the two satisfaction categories.

The resulting confusion matrix was:

| | Predicted Neutral/Dissatisfied | Predicted Satisfied |
|---|---:|---:|
| **Actual Neutral/Dissatisfied** | 2,792 | 123 |
| **Actual Satisfied** | 194 | 2,087 |

The results can be interpreted as follows:

- **True Negatives: 2,792** — neutral or dissatisfied passengers correctly identified.
- **False Positives: 123** — passengers predicted as satisfied but actually neutral or dissatisfied.
- **False Negatives: 194** — passengers predicted as neutral or dissatisfied but actually satisfied.
- **True Positives: 2,087** — satisfied passengers correctly identified.

### Key Finding

The Decision Tree correctly classified a large majority of passengers in both satisfaction categories while making relatively few incorrect predictions.

This supports the model's strong testing performance of **93.90% accuracy**.

### Business Interpretation

The confusion matrix suggests that the Decision Tree can effectively distinguish between satisfied and neutral or dissatisfied passengers.

The model could therefore be used as a decision-support tool to identify satisfaction patterns and help the airline investigate passengers or service areas that may require attention.

However, the model should not be treated as a perfect predictor of individual passenger satisfaction. Its predictions should be considered alongside business knowledge and other passenger information.

## 4.11 Model Comparison

Two classification models were evaluated:

1. Logistic Regression
2. Decision Tree

Both models were evaluated using the same testing dataset and the same performance metrics.

| Model | Accuracy | Precision | Recall | F1-score |
|---|---:|---:|---:|---:|
| Logistic Regression | 87.24% | 86.18% | 84.48% | 85.32% |
| **Decision Tree** | **93.90%** | **94.43%** | **91.49%** | **92.94%** |

### Key Finding

The Decision Tree outperformed Logistic Regression across all four evaluation metrics.

Compared with Logistic Regression, the Decision Tree achieved:

- **6.66 percentage points higher accuracy**
- **8.25 percentage points higher precision**
- **7.01 percentage points higher recall**
- **7.62 percentage points higher F1-score**

### Business Interpretation

The Decision Tree provides stronger predictive performance than Logistic Regression for this dataset.

This suggests that the relationships between passenger characteristics, travel information, service ratings, and satisfaction may be better captured by a Decision Tree than by the simpler Logistic Regression model.

---

## 4.12 Checking for Overfitting

A model can perform very well on training data but perform poorly on new data. This is known as **overfitting**.

To check for overfitting, the Decision Tree's training and testing accuracy were compared.

- **Training Accuracy:** 95.34%
- **Testing Accuracy:** 93.90%
- **Difference:** approximately 1.44 percentage points

### Key Finding

The difference between training and testing accuracy is relatively small.

This suggests that the Decision Tree is generalizing well to unseen passenger data and does not show strong evidence of overfitting.

### Business Interpretation

The Decision Tree is not simply memorizing the passengers in the training dataset.

Its strong performance on the unseen testing dataset suggests that it has learned patterns that can also be applied to new passenger records.

This increases confidence in using the model as a decision-support tool.
